## NumPy — Numerical Calculations

**Definition:** NumPy (Numerical Python) is a powerful Python library used for working with multi-dimensional arrays and matrices. It provides highly efficient mathematical functions to perform operations on these arrays without needing slow python loops.

Here are 4 simple, beginner-friendly examples demonstrating NumPy in action.

```python
import numpy as np

print("--- NumPy Example 1: Creating an Array and Calculating Average ---")
# 1. Store student exam scores in a NumPy array
exam_scores = np.array([85, 90, 78, 92, 88])

# 2. Calculate the average score using np.mean()
average_score = np.mean(exam_scores)

print("Exam Scores:", exam_scores)
print("Average Exam Score:", average_score)
print()
```

```python
print("--- NumPy Example 2: Element-wise Arithmetic (Adding a Bonus) ---")
# 1. Base scores of students
base_scores = np.array([70, 75, 80, 85])

# 2. Add a flat bonus of 5 points to every student at once
final_scores = base_scores + 5

print("Base Scores:", base_scores)
print("Final Scores (with +5 bonus):", final_scores)
print()
```

```python
print("--- NumPy Example 3: Finding Maximum and Minimum Values ---")
# 1. Monthly sales figures in dollars
monthly_sales = np.array([1200, 1500, 900, 2100, 1800])

# 2. Find highest and lowest monthly sales
max_sales = np.max(monthly_sales)
min_sales = np.min(monthly_sales)

print("Monthly Sales Data:", monthly_sales)
print("Highest Sale:", max_sales)
print("Lowest Sale:", min_sales)
print()
```

```python
print("--- NumPy Example 4: Filtering Data (Conditional Selection) ---")
# 1. Temperature readings in Celsius
temperatures = np.array([16, 22, 14, 31, 25, 12])

# 2. Filter temperatures that are above 20 degrees Celsius
hot_days = temperatures[temperatures > 20]

print("All Temperatures:", temperatures)
print("Temperatures above 20°C:", hot_days)
print()
```

## Pandas — Data Preparation and Analysis

**Definition:** Pandas is a highly popular Python library used for data manipulation and analysis. It introduces the **DataFrame**, which organizes data into a structured table with labeled rows and columns (just like an Excel spreadsheet or SQL table), making it incredibly easy to clean, filter, and inspect datasets.

Here are 4 simple, practical examples demonstrating Pandas in action.

```python
import pandas as pd

print("--- Pandas Example 1: Creating a DataFrame ---")
# 1. Prepare data inside a Python dictionary
data = {
    "Student": ["Sophia", "Liam", "Emma", "Noah"],
    "Score": [92, 78, 95, 84],
    "Department": ["Science", "Art", "Science", "Math"]
}

# 2. Convert the dictionary into a Pandas DataFrame
df = pd.DataFrame(data)

# 3. Display the DataFrame
display(df)
print()
```

```python
print("--- Pandas Example 2: Filtering Rows by Condition ---")
# From our previous DataFrame, filter out students who scored 85 or above
high_scorers = df[df["Score"] >= 85]

display(high_scorers)
print()
```

```python
print("--- Pandas Example 3: Adding a New Calculated Column ---")
# Add a boolean column 'Passed' that checks if a student's score is >= 80
df["Passed"] = df["Score"] >= 80

display(df)
print()
```

```python
print("--- Pandas Example 4: Grouping and Calculating Statistics ---")
# Group students by their department and calculate the average score for each department
dept_averages = df.groupby("Department")["Score"].mean().reset_index()

display(dept_averages)
print()
```

## Matplotlib and Seaborn — Data Visualization

**Definition:** Matplotlib and Seaborn are two of Python's most powerful libraries for data visualization. Matplotlib provides the foundation for building charts, while Seaborn acts as an extension to make them look modern, beautiful, and clean with minimal code.

Here are 4 distinct, visually polished, and easy-to-understand visualization examples.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set a clean visual theme for all plots
sns.set_theme(style="whitegrid")

print("--- Visualization Example 1: Bar Chart (Comparing Student Scores) ---")
# 1. Prepare data
students = ["Sophia", "Liam", "Emma", "Noah"]
scores = [92, 78, 95, 84]

# 2. Create a clean bar plot
plt.figure(figsize=(6, 4))
sns.barplot(x=students, y=scores, hue=students, palette="Blues_d", legend=False)

# 3. Add labels and title
plt.title("Comparison of Student Exam Scores", fontsize=14, pad=15)
plt.xlabel("Students", fontsize=12)
plt.ylabel("Exam Score (out of 100)", fontsize=12)
plt.ylim(0, 100)

# 4. Show the chart cleanly
plt.tight_layout()
plt.show()
```

```python
print("--- Visualization Example 2: Line Chart (Tracking Score Trends over Weeks) ---")
# 1. Prepare data
weeks = ["Week 1", "Week 2", "Week 3", "Week 4"]
average_grades = [75, 80, 78, 88]

# 2. Plot score trends with custom markers
plt.figure(figsize=(6, 4))
plt.plot(weeks, average_grades, marker='o', color='#2ca02c', linewidth=2.5, markersize=8)

# 3. Customize aesthetics
plt.title("Class Average Progression Over Weeks", fontsize=14, pad=15)
plt.xlabel("Time (Weeks)", fontsize=12)
plt.ylabel("Average Grade (%)", fontsize=12)
plt.ylim(60, 100)

plt.tight_layout()
plt.show()
```

```python
print("--- Visualization Example 3: Scatter Plot (Study Hours vs. Exam Scores) ---")
# 1. Prepare data
study_hours = [2, 4, 5, 7, 8, 10, 12]
student_scores = [55, 65, 70, 78, 85, 90, 98]

# 2. Create the scatter plot to see correlation
plt.figure(figsize=(6, 4))
sns.scatterplot(x=study_hours, y=student_scores, color="#ff7f0e", s=100, edgecolor="black")

# 3. Add details
plt.title("Effect of Study Hours on Exam Scores", fontsize=14, pad=15)
plt.xlabel("Hours Studied", fontsize=12)
plt.ylabel("Final Exam Score", fontsize=12)
plt.xlim(0, 14)
plt.ylim(40, 105)

plt.tight_layout()
plt.show()
```

```python
print("--- Visualization Example 4: Box Plot (Distribution of Department Scores) ---")
# 1. Prepare a DataFrame for more structured plotting
df_viz = pd.DataFrame({
    "Department": ["Science", "Science", "Math", "Math", "Art", "Art", "Art"],
    "Score": [95, 92, 84, 88, 78, 81, 75]
})

# 2. Create a clean box plot showing distributions
plt.figure(figsize=(6, 4))
sns.boxplot(data=df_viz, x="Department", y="Score", hue="Department", palette="Set2", legend=False)

# 3. Set titles
plt.title("Distribution of Exam Scores by Department", fontsize=14, pad=15)
plt.xlabel("Department", fontsize=12)
plt.ylabel("Score Distribution", fontsize=12)

plt.tight_layout()
plt.show()
```

## Scikit-learn — Machine Learning

**Definition:** Scikit-learn is the standard Python library used for Machine Learning. It provides clean and consistent tools for modeling, data preprocessing, splits, training algorithms, and calculating performance metrics.

Here are 4 clear, essential, and beginner-friendly examples to demonstrate the typical machine learning workflow.

```python
from sklearn.model_selection import train_test_split
import numpy as np

print("--- ML Example 1: Splitting Data into Train and Test Sets ---")
# 1. Simulate dataset: Study Hours (Features) and Grades (Targets)
X = np.array([[2], [4], [6], [8], [10], [12], [14], [16]])
y = np.array([55, 60, 68, 75, 80, 85, 90, 95])

# 2. Split dataset: 75% for training, 25% for testing
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)

print("Training Features (Study Hours):\n", X_train)
print("Testing Features (Study Hours):\n", X_test)
print()
```

```python
from sklearn.linear_model import LinearRegression

print("--- ML Example 2: Training a Simple Regression Model ---")
# 1. Initialize the Linear Regression model
model = LinearRegression()

# 2. Train (fit) the model on our training data
model.fit(X_train, y_train)

print("Model trained successfully!")
print(f"Formula learned: Grade = {model.coef_[0]:.2f} * Hours + {model.intercept_:.2f}")
print()
```

```python
print("--- ML Example 3: Making Predictions on New Data ---")
# Predict the score for a student who studied 5 hours and another who studied 11 hours
new_students = np.array([[5], [11]])
predicted_grades = model.predict(new_students)

for hours, grade in zip(new_students.flatten(), predicted_grades):
    print(f"Studied {hours} hours -> Predicted Grade: {grade:.2f}%")
print()
```

```python
from sklearn.metrics import mean_squared_error, r2_score

print("--- ML Example 4: Evaluating Model Performance ---")
# 1. Generate predictions on our reserved test dataset
test_predictions = model.predict(X_test)

# 2. Calculate performance metrics
mse = mean_squared_error(y_test, test_predictions)
r2 = r2_score(y_test, test_predictions)

print("Actual Grades:", y_test)
print("Predicted Grades:", np.round(test_predictions, 2))
print(f"Mean Squared Error (MSE): {mse:.2f} (Lower is better)")
print(f"R-squared Score (Accuracy): {r2:.2f} (Closer to 1.0 is better)")
print()
```
